# Setup a FloodAdapt Database

This notebook allows a user to setup SFINCS, WFLOW, Delft-FIAT, and RA2CE models and configure them into a FloodAdapt Database. The individual model setups are based on global data hosted on the InterTwin Datalake. The user provides a minimal set of inputs for the model setup. For more detailed configurations, the model setups are based on the config files found in the DT-flood repo under DT_flood/cfg. \
For both fetching the data and running the worklow, the right credentials are necessary. This is handled through EGI SSO, where the user should be a member of the dev.intertwin.eu VO

## **Step 0:** Import Dependencies

```python
from pathlib import Path

from DT_flood.utils.plot_utils import create_base_map
from DT_flood.utils.workflow_utils import create_setup_workflow_config, run_setup_database_workflow
```

## **Step 1:** Define area of interest

The cell below opens an interactive map where the user can draw the bounds of the region of interest. This is stored in the `selected_geometry` variable. Note that a valid geometry has to be a closed polygon. \
Alternatively, a region of interest can be specified through a user-provided geojson file (see below.)

```python
m ,selected_geometry = create_base_map()
m
```

## **Step 2:** Workflow inputs

Here the inputs to the workflow are defined. These are:
- `database_path`: Location where the final FloodAdapt Database will be saved
- `region_file`: Definition of the area of interest. Either the `selected_geometry` var from above, or a Path to a geojson file
- `sf_res`: Resolution of the SFINCS model
- `sf_nr_subpixels`: Number of subgrid pixels in SFINCS model
- `wf_res`: Resolution of WFLOW model
- `ra2ce_dest_points`: OSM tag used to get RA2CE destination points, by default 'hopsital'

```python
database_path = Path("/home/wotromp/InterTwin/workflow_refactor/model_setup/databases/test_database")

region_file = Path("/home/wotromp/InterTwin/workflow_refactor/model_setup/HumberDelta_large.geojson")
# region_file = selected_geometry

oscar_endpoint = 'https://oscar-ukri.intertwin.fedcloud.eu'
oscar_token = ""
```

## **Step 3:** Launch workflow

The cell below will create the config file for the CWL workflow. This is used one cell after to launch the workflow itself. The `debug` argument toggles whether jobs are cached or not. \
*NB: Make sure Rucio is authenticated for data fetching by running `rucio whoami` on the command line.*

```python
create_setup_workflow_config(
    database_path=database_path,
    region=region_file,
    oscar_endpoint=oscar_endpoint,
    oscar_token=oscar_token
)
```

```python
run_setup_database_workflow(
    database_path=database_path,
    debug=True
)
```
